# Build Embedding

```python
#git checkout -b feat/env-vir
```

```python
from sentence_transformers import SentenceTransformer
```

```python
Model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
docs =[
    'I love AI',
    'The weather is amazing today',
    'ML is cool'
]
```

```python
embeddings = Model.encode(docs)
```

```python
embeddings.shape #3 sentence + 384 features
```

# Query

```python
query = "Atrifical Intelligence"
```

```python
query_embedding = Model.encode([query])
```

```python
print(query_embedding.shape)
```

# Similarity 

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
score = cosine_similarity(query_embedding,embeddings)
print(score)
```

```python
best_match_index = score.argmax()
```

```python
best_match_index
```

```python
answer = docs[best_match_index]
answer
```
